```python
import math
```

```python
L1 = float(input("Enter length of first arm (L1): "))
L2 = float(input("Enter length of second arm (L2): "))

theta1 = float(input("Enter first angle (theta1 in degrees): "))
theta2 = float(input("Enter second angle (theta2 in degrees): "))
```

```python
t1 = math.radians(theta1)
t2 = math.radians(theta2)
```

```python
x = L1 * math.cos(t1) + L2 * math.cos(t1 + t2)
y = L1 * math.sin(t1) + L2 * math.sin(t1 + t2)
```

```python
print("Robot Arm Position:")
print("X =", round(x, 2))
print("Y =", round(y, 2))
```
